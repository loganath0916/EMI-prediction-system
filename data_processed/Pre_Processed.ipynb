```python
#Clean Numeric Columns
df['age'] = pd.to_numeric(df['age'], errors='coerce')

df['monthly_salary'] = pd.to_numeric(
    df['monthly_salary'],
    errors='coerce'
)

df['bank_balance'] = pd.to_numeric(
    df['bank_balance'],
    errors='coerce'
)
```

```python
#Missing Value Report

missing = df.isnull().sum()

missing_percent = (
    missing / len(df)
) * 100

pd.DataFrame({
    'Missing_Count': missing,
    'Missing_Percent': missing_percent
}).sort_values(
    'Missing_Count',
    ascending=False
)
```

```python
#fill categorical column
df['education'] = df['education'].fillna(
    df['education'].mode()[0]
)
```

```python
#fill numerical column
num_cols = [
    'monthly_rent',
    'credit_score',
    'bank_balance',
    'emergency_fund',
    'monthly_salary',
    'age'
]

for col in num_cols:
    df[col] = df[col].fillna(
        df[col].median()
    )
```

```python
#Perform Final Data Validation
print("Duplicates:", df.duplicated().sum())

print("Age < 25:", (df['age'] < 25).sum())
print("Age > 60:", (df['age'] > 60).sum())

print("Credit < 300:", (df['credit_score'] < 300).sum())
print("Credit > 850:", (df['credit_score'] > 850).sum())

print("Salary < 15000:", (df['monthly_salary'] < 15000).sum())
print("Salary > 200000:", (df['monthly_salary'] > 200000).sum())
```

```python
df = df[
    (df['credit_score'] >= 300) &
    (df['credit_score'] <= 850)
]

df = df[
    (df['monthly_salary'] >= 15000) &
    (df['monthly_salary'] <= 200000)
]
```
