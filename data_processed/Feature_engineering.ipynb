```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv(r"C:\Users\HP\OneDrive\Desktop\python\guvi_pro\EMI prediction\data\EMI_dataset_cleaned.csv")
```

```python
#Debt-to-Income Ratio (DTI)
df['debt_to_income_ratio'] = (
    df['current_emi_amount'] /
    df['monthly_salary']
)
```

```python
#Expense-to-Income Ratio
df['total_expenses'] = (
    df['school_fees'] +
    df['college_fees'] +
    df['travel_expenses'] +
    df['groceries_utilities'] +
    df['other_monthly_expenses'] +
    df['monthly_rent']
)

df['expense_to_income_ratio'] = (
    df['total_expenses'] /
    df['monthly_salary']
)
```

```python
#Affordability Ratio
df['affordability_ratio'] = (
    (
        df['monthly_salary']
        - df['total_expenses']
        - df['current_emi_amount']
    )
    /
    df['monthly_salary']
)
```

```python
#Credit Risk Score
def credit_risk(score):

    if score >= 750:
        return "Low"

    elif score >= 650:
        return "Medium"

    else:
        return "High"


df['credit_risk'] = df['credit_score'].apply(
    credit_risk
)
```

```python
#Employment Stability Score
def employment_stability(years):

    if years >= 10:
        return "High"

    elif years >= 3:
        return "Medium"

    else:
        return "Low"


df['employment_stability'] = (
    df['years_of_employment']
    .apply(employment_stability)
)
```

```python
#Salary × Credit Score
df['salary_credit_interaction'] = (
    df['monthly_salary']
    *
    df['credit_score']
)
```

```python
#Salary × Employment Experience
df['salary_experience_interaction'] = (
    df['monthly_salary']
    *
    df['years_of_employment']
)
```

```python
#Categorical Encoding
le = LabelEncoder()

categorical_cols = [
    'gender',
    'marital_status',
    'education',
    'employment_type',
    'company_type',
    'house_type',
    'existing_loans',
    'emi_scenario',
    'credit_risk',
    'employment_stability'
]

for col in categorical_cols:
    df[col] = le.fit_transform(df[col])
```

```python
df['emi_eligibility'] = (
    LabelEncoder()
    .fit_transform(df['emi_eligibility'])
)
```
